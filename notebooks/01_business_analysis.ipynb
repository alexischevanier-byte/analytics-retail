```python
region_por_cliente = df.groupby('customer_id')['region'].nunique().loc[lambda x: x > 1]
region_por_cliente
```

```python
region_con_nulo = df.loc[df['region'].isna(), 'customer_id'].unique()

region_por_cliente = df.groupby('customer_id')['region'].nunique()

region_por_cliente.loc[region_con_nulo]
```

```python
region_cliente = df.groupby('customer_id')['region'].first()
region_cliente.head()
```

```python
df['region'] = df['region'].fillna(df['customer_id'].map(region_cliente))
df['region'].isna().sum()
```

```python
metodos_de_pago = df.groupby('customer_id')['payment_method'].nunique().loc[lambda x: x > 1]
metodos_de_pago
```

```python
df['payment_method'] = df['payment_method'].fillna('Unknown')
df.isna().sum()
```
